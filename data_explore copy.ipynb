```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_slip = pd.read_csv("data/friction_data/features_AgingLaw_v2.csv").values
Y_slip = pd.read_csv("data/friction_data/targets_AgingLaw_v2.csv").values

fig, axes = plt.subplots(2,2,figsize=(10,8))
axes = axes.flatten()

for i in range(4):
    axes[i].plot([j/X_slip.shape[1] for j in range(X_slip.shape[1])], X_slip[i])
    axes[i].set_yscale('log')

    twin_axes = axes[i].twinx()
    twin_axes.plot([j/X_slip.shape[1] for j in range(Y_slip.shape[1])], Y_slip[i], color='orange')
    twin_axes.set_ylabel("Output Friction", color='orange', fontsize = 16)
    twin_axes.tick_params(axis='y', labelcolor='orange', labelsize = 16)
    axes[i].set_xlabel(f"Time (s)", fontsize = 16)
    axes[i].set_ylabel("Input Velocity (m/s)", color='blue', fontsize = 16)
    axes[i].tick_params(axis='y', labelcolor='blue', labelsize= 16)

plt.tight_layout()
```

```python
X_age = pd.read_csv("data/friction_data/features_AgingLaw_v2.csv").values
Y_age = pd.read_csv("data/friction_data/targets_AgingLaw_v2.csv").values

fig, axes = plt.subplots(3,3,figsize=(15,15))
axes = axes.flatten()

for i in range(9):
    axes[i].plot([j for j in range(X_age.shape[1])], X_age[i])

    twin_axes = axes[i].twinx()
    twin_axes.plot([j for j in range(Y_age.shape[1])], Y_age[i], color='red')
    twin_axes.set_ylabel("Output Friction", color='red')
    twin_axes.tick_params(axis='y', labelcolor='red')
    axes[i].set_xlabel(f"Time")
    axes[i].set_ylabel("Input Velocity", color='blue')
    axes[i].tick_params(axis='y', labelcolor='blue')

plt.tight_layout()
```
